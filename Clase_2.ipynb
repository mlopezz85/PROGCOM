##For
Usada cuando se sabe cuántas veces se debe repetir una acción
for(variable) in range(VI,VF+1, Paso):
   Se repite (VF-VI)/Paso veces. VF infica la cantidad de veces que se desea repetir una acción.

```python
#Imrpimir múltiplos de 3, hasta que se encuentren del 1 al 30
for intento in range(31): #el número entre paréntesis se refiere al rango de los datos en los que se mira. El valor final es siempre +1
  #establecer quienes son múltiplos de 3

  if intento%3==0:
    print(intento)
```

```python
for prueba in range(3,31,3): #Esto hace que no se tome el 0 como múltiplo de 3. El último 3 indica los pasos que se dan, es decir, suma de 3 en 3. Esta indicación hace lo mismo. LOs primeros dos números son el rango.
  print(prueba)
```

```python
#Despegue de un cohete
for despegue in range(10,0,-1): #Como es cuenta regresiva, el rango se pone al revés y los pasos se indican omo -1
  print(despegue)
```

```python
estado=True
opcion=0
actividad=""
actividades=[]
c=0
buscar=0
cantidad=0
prioridad=""
while estado:
  print("¡¡BIENVENIDO/A!! ¿Qué deseas hacer?")
  print("1. Ver agenda")
  print("2. Agregar actividad")
  print("3. Buscar actividad")
  print("4. Eliminar actividad")
  print("5. Mostrar cantidad de actividades")
  print("6. Ver primera actividad")
  print("7. Ver última actividad")
  print("8. Ordenar agenda alfabéticamente")
  print("9. Salir")
  opcion=int(input("¿Qué desea hacer? "))
  match opcion:
    case 1:
      print("---AGENDA---")
      print(f"\nLa agenda contiene: {actividades}\n")
    case 2:
      print("---AGREGAR ACTIVIDADES---")
      c=int(input("¿Cuántas actividades desea agregar? "))
      for i in range(c):
        actividad=input("Escriba la actividad que desea añadir a la agenda: ")
        actividades.append(actividad)
        prioridad=str("¿Qué prioridad tiene la actividad? {baja, media, alta}")
        print("Actividad añadida con éxito :) \n")
    case 3:
      print("---BUSCAR ACTIVIDADES---")
      buscar=int(input("¿Qué actividad deseas buscar?"))
      print(f"Aquí tienes: \n{actividades[buscar]}\n")
    case 4:
      print("---ELIMINAR ACTIVIDADES---")
      eliminar=int(input("¿Qué actividad deseas eliminar?"))
      actividades.pop(eliminar)
      print("Actividad eliminada con éxito :)\n")
    case 5:
      print("---MOSTRAR CANTIDAD DE ACTIVIDADES---")
      cantidad=len(actividades)
      print(f"La agente tiene {cantidad} de actividades pendientes.\n")
    case 6:
      print("---VER PRIMERA ACTIVIDAD---")
      print(f"La primera actividad en la agenda es: {actividades[0]}\n")
    case 7:
      print("---VER ÚLTIMA ACTIVIDAD---")
      print(f"La última actividad en la agenda es: {actividades[-1]}\n")#el -1 indica el último elemento en la lista, y así sucesivamente. El -2 sería el penúltimo y así.
    case 8:
      print("---ORDENAR AGENDA ALFABÉTICAMENTE---")
      if not actividades:
        print("\nLa agenda está vacía, no hay nada que ordenar :.\n")
      else:
        actividades = sorted(actividades)
        print("Se ha ordenado la agenda con éxito :)\n")
    case 9:
      estado=False
      print("¡Hasta luego!")
    case _:
      print("\nOpción no válida, por favor intente de nuevo.\n")
```

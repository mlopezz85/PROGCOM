```python
#Ejercicio:  forma 1
nombre=["sofia","karen","lucila","julian","mayra"]
filtro=[]
for n in nombre: #iterador
  if n[0]=="m": filtro.append(n)
filtro
```

```python
#Forma 2
f2=[n for n in nombre if n[0]=="m"]
f2
```

```python
#Para mostrar los elementos con más longitud
f3=[n for n in nombre if len(n)>=5]
f3
```

```python
#Imprimir los nombres de los estudiantes que pasaron el parcial
notas=[2.5,3.2,0.1,4.9,3]
f4=[z for z in notas if z>=3]
f4
```

```python
f5=[nombre[ye] for ye in range(5) if notas[ye]>=3] #if nota... es la iteración que según la repetición que se haga, va a mirar cada elemento dentro de la lista
f5
```

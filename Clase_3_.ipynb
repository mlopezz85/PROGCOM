##Estructura mientras (while)**

(En Java no existe la función de "hasta que se cumpla (condición)"
Mientras que la condición sea verdadera, se repetira n veces



```

   Ejecuta el código
```

````python
#Tabla de multiplicar del 5
print("Tabla del 5")
i=1
while i<=10: #el 10 significa hasta qué número se va a multiplicar el 5
  print(f"5 x {i} = {i*5}")
  i=i+1
````

````python
#Tanda 3 -
i=1
m=1
while i<=10:
  print(f"Tabla de multiplicar de {i}")
  while  m<=10:
    print(f"{i} x {m} = {i*m}")
    m+=1
````

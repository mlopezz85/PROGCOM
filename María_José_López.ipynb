##if/elif/else

Evalua múltiples condiciones como elif hayan. Si no se cumple, se verifica la siguiente condición, y así sucesivamente hasta que se encuentre un resultado.

Sintáxis

if (condicion1):
   ejecución
if (condicion2):
   ejecución
else:
   ejecución

```python
edad=""
edad=int(input("Ingrese su edad. "))
if (edad>0 and edad <18):
  print("Menor de edad")
elif (edad>=18 and edad<60):
  print("Mayor de edad.")
else:
  print("Adulto mayor.                                                                                                                                                                                                          ")
```

##match/case
La estructura match evalúa una variable frente a varios bloques case.
El comodín case_ funciona como "ninguna de las anteriores"

Sintáxis

match(condicion):
  casev1=ejecuta si condición==v1
  casev2=ejecuta si condición==v2
  casev3=ejecuta si condición==v3
  casev4   =ejecuta si condición==v4

```python
opcion=""
int(input("Ingrese una opción del 1 al 3. "))
match opcion:
  case "1":
    print("Opción 1")
  case "2":
    print("Opción 2")
  case "3":
    print("Opción 3")
```

#Saber si el día es caliente, templado o frío
* Caliente-> 26+
* Templado->20-26
* Frío-> <20

```python
temperatura=int(input("Ingrese la temperatura del ambiente. "))
if temperatura>=26:
  print("¡¡QUÉ CALOR!!")
elif temperatura<26 and temperatura>=20:
  print("Qué tíbios.")
else:
  print("¡¡QUÉ FRÍO!!")
```
